```python
#2.Q2. Create a DataFrame and find the mean salary grouped by department.
{'Name': ['Amit', 'Sneha', 'Kunal', 'Riya'], 'Department': ['HR', 'IT', 'HR', 'IT'], 'Salary': [30000, 50000, 35000, 52000]}
Expected Output:
Department
HR    32500.0
IT    51000.0
Name: Salary, dtype: float64
```

```python
import pandas as pd
data={'Name': ['Amit', 'Sneha', 'Kunal', 'Riya'], 'Department': ['HR', 'IT', 'HR', 'IT'], 'Salary': [30000, 50000, 35000, 52000]}
dataset=pd.DataFrame(data)
total=dataset.groupby('Department')['Salary'].mean()
total
```

```python
#3.Q3. Create a DataFrame and apply label encoding to the Gender column.
{'Name': ['Arjun', 'Leela', 'Kiran'], 'Gender': ['Male', 'Female', 'Male']}
Expected Output:
    Name  Gender  Gender_encoded
0  Arjun    Male               1
1  Leela  Female               0
2  Kiran    Male               1
```

```python
import pandas as pd
data={'Name': ['Arjun', 'Leela', 'Kiran'], 'Gender': ['Male', 'Female', 'Male']}
dataset=pd.DataFrame(data)
dataset['Gender_encoded']=pd.get_dummies(dataset['Gender'],drop_first=True)
dataset
```

```python
#4.🔹Q4. Normalize the 'Score' column using Min-Max scaling.
{'Student': ['A', 'B', 'C'], 'Score': [55, 80, 65]}

Expected Output:
    Student  Score  Normalized_Score
0       A     55               0.0
1       B     80               1.0
2       C     65               0.4
```

```python
import pandas as pd
data={'Student': ['A', 'B', 'C'], 'Score': [55, 80, 65]}
dataset=pd.DataFrame(data)
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
dataset['Normalized_Score']=scaler.fit_transform(dataset[['Score']])
dataset
```

```python
#5.
Use SVM, RandomForest, and GaussianNB to classify customers as 'Buy' or 'Not Buy' based on Age and Salary.
{'Age': [22, 25, 47, 52, 46, 56, 23], 'Salary': [15000, 29000, 48000, 60000, 52000, 61000, 17000], 'Buy': [0, 0, 1, 1, 1, 1, 0]}
```

```python
import pandas as pd
data = {'Age': [22, 25, 47, 52, 46, 56, 23], 'Salary': [15000, 29000, 48000, 60000, 52000, 61000, 17000], 'Buy': [0, 0, 1, 1, 1, 1, 0]}
dataset=pd.DataFrame(data)
dataset


independent=dataset[['Age','Salary']]
dependent=dataset[['Buy']]
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.3)

from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix,classification_report

#SVC
classifier=SVC(C=10)
model=classifier.fit(x_train,y_train)
y_pred=model.predict(x_test)
cm=confusion_matrix(y_test,y_pred)
report = classification_report(y_test,y_pred)
print("SVM:")
print(cm)
print(report)

#RandomForestClassifier
classifier=RandomForestClassifier()
model=classifier.fit(x_train,y_train)
y_pred=model.predict(x_test)
cm=confusion_matrix(y_test,y_pred)
report = classification_report(y_test,y_pred)
print("RandomForestClassifier:")
print(cm)
print(report)

#GaussianNB
classifier=GaussianNB()
model=classifier.fit(x_train,y_train)
y_pred=model.predict(x_test)
cm=confusion_matrix(y_test,y_pred)
report = classification_report(y_test,y_pred)
print("GaussianNB:")
print(cm)
print(report)
```
