```python
import pickle
import joblib
filename="final_model.sav"
#load model and scaler
loaded_inputscaler=joblib.load(open("input.joblib",'rb'))
loaded_outputscaler=joblib.load(open("output.joblib",'rb'))
loaded_model=pickle.load(open(filename,'rb'))
#preprocessing input
preinput=loaded_inputscaler.transform([[1200,1200,34,0,0]])
#predict
preoutput=loaded_model.predict(preinput)
preoutput
#preprocessing output
output=loaded_outputscaler.inverse_transform([preoutput])
output
```
