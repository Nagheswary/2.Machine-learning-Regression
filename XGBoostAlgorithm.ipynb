```python
from xgboost import XGBRegressor
```

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
dataset=pd.read_csv("insurance_pre.csv")
encoder=OrdinalEncoder(categories=[['female','male']])
dataset['sex']=encoder.fit_transform(dataset[['sex']])
encoder=OrdinalEncoder(categories=[['yes','no']])
dataset['smoker']=encoder.fit_transform(dataset[['smoker']])
independent=dataset[['age', 'sex', 'bmi', 'children', 'smoker']]
dependent=dataset[['charges']]
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
regressor=XGBRegressor(objective ='reg:linear', n_estimators = 10,max_depth=5, learning_rate=0.1)
regressor.fit(x_train,y_train)
y_pred=regressor.predict(x_test)
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
r_score
```
